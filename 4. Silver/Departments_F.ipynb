```python
from pyspark.sql import SparkSession, functions as f

#Reading Hospital A departments data 
df_hosa=spark.read.parquet("abfss://bronze@ttadlsjcrh.dfs.core.windows.net/hosa/departments")

#Reading Hospital B departments data 
df_hosb=spark.read.parquet("abfss://bronze@ttadlsjcrh.dfs.core.windows.net/hosb/departments")

#union two departments dataframes
df_merged = df_hosa.unionByName(df_hosb)

# Create the dept_id column and rename deptid to src_dept_id
df_merged = (df_merged.withColumn("SRC_Dept_id", f.col("deptid"))
                     .withColumn("Dept_id", f.concat(f.col("deptid"), f.lit('-'), f.col("datasource")))
                     .drop("deptid")
                     )

df_merged.createOrReplaceTempView("departments")
```

```python
%sql
CREATE TABLE IF NOT EXISTS silver.departments (
Dept_Id string,
SRC_Dept_Id string,
Name string,
datasource string,
is_quarantined boolean
)
USING DELTA;
```

```python
%sql 
truncate table silver.departments
```

```python
%sql
insert into silver.departments
SELECT 
Dept_Id,
SRC_Dept_Id,
Name,
Datasource,
    CASE 
        WHEN SRC_Dept_Id IS NULL OR Name IS NULL THEN TRUE
        ELSE FALSE
    END AS is_quarantined
FROM departments
```

```python
df_sil_departments = spark.read.format("delta").table("silver.departments")

# Define your ADLS Gen2 path
adls_path = "abfss://silver@ttadlsjcrh.dfs.core.windows.net/"

# Write the DataFrame as a Delta table
df_sil_departments.write.format("delta").mode("overwrite").save(adls_path + "/departments")
```
