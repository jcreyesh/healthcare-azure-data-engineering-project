```python
%sql
CREATE TABLE IF NOT EXISTS gold.dim_npi (
  npi_id STRING,
  first_name STRING,
  last_name STRING,
  position STRING,
  organisation_name STRING,
  last_updated STRING,
  refreshed_at TIMESTAMP)
```

```python
%sql
-- truncate table gold.dim_npi
```

```python
%sql
insert into
  gold.dim_npi
select
  npi_id,
  first_name,
  last_name,
  position,
  organisation_name,
  last_updated,
  current_timestamp()
from
  silver.npi_extract
  where is_current_flag = true
```

```python
%python
df_gold_dim_npi = spark.read.format("delta").table("gold.dim_npi")

# Define your ADLS Gen2 path
adls_path = "abfss://gold@ttadlsjcrh.dfs.core.windows.net/"

# Write the DataFrame as a Delta table
df_gold_dim_npi.write.format("delta").mode("overwrite").save(adls_path + "/npi_extract")
```
