```python
%sql
CREATE TABLE IF NOT EXISTS gold.dim_provider
(
ProviderID string,
FirstName string,
LastName string,
DeptID string,
NPI long,
datasource string
)
```

```python
%sql 
-- truncate TABLE gold.dim_provider
```

```python
%sql
insert into gold.dim_provider
select 
ProviderID ,
FirstName ,
LastName ,
concat(DeptID,'-',datasource) deptid,
NPI ,
datasource 
 from silver.providers
 where is_quarantined=false
```

```python
df_gol_dim_providers = spark.read.format("delta").table("gold.dim_provider")

# Define your ADLS Gen2 path
adls_path = "abfss://gold@ttadlsjcrh.dfs.core.windows.net/"

# Write the DataFrame as a Delta table
df_gol_dim_providers.write.format("delta").mode("overwrite").save(adls_path + "/providers")
```
