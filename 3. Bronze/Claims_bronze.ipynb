```python
from pyspark.sql import SparkSession, functions as f

claims_df = spark.read.csv("abfss://landing@ttadlsjcrh.dfs.core.windows.net/claims/*.csv",header=True)

claims_df = claims_df.withColumn(
    "datasource",
    f.when(f.col('_metadata.file_name').contains("hospital1"), "hosa").when(f.col('_metadata.file_name').contains("hospital2"), "hosb")
     .otherwise(None)
)

display(claims_df.limit(10))
```

```python
claims_df.write.format("parquet").mode("overwrite").save("abfss://bronze@ttadlsjcrh.dfs.core.windows.net/claims")
```
